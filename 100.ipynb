```python
# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# 2. Load Dataset

data = pd.read_csv("/content/car_price_dataset.csv")

print("=" * 60)
print("CAR PRICE PREDICTION")
print("=" * 60)

print("\nFirst 5 Records:")
print(data.head())

# 3. Display Dataset Information

print("\nDataset Shape:")
print(data.shape)

print("\nColumn Names:")
print(data.columns.tolist())

print("\nDataset Information:")
print(data.info())

print("\nStatistical Description:")
print(data.describe())

# 4. Check Missing Values

print("\nMissing Values:")
print(data.isnull().sum())

# 5. Check Duplicate Values

print("\nNumber of Duplicate Rows:")
print(data.duplicated().sum())

# Remove duplicate rows if present
data = data.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(data.shape)

# 6. Exploratory Data Analysis

# Distribution of selling price

plt.figure(figsize=(9, 5))

plt.hist(
    data["Selling_Price"],
    bins=10,
    edgecolor="black"
)

plt.xlabel("Selling Price (Lakhs)")
plt.ylabel("Number of Cars")
plt.title("Distribution of Car Selling Prices")

plt.grid(axis="y", alpha=0.3)

plt.show()

# 7. Present Price vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Present_Price"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Present Price (Lakhs)")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Present Price vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 8. Year vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Year"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Year vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 9. Kms Driven vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Kms_Driven"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Kilometers Driven vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 10. Fuel Type Analysis

plt.figure(figsize=(7, 5))

sns.boxplot(
    x="Fuel_Type",
    y="Selling_Price",
    data=data
)

plt.xlabel("Fuel Type")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Selling Price by Fuel Type")

plt.show()

# 11. Encode Categorical Variables

fuel_encoder = LabelEncoder()
seller_encoder = LabelEncoder()
transmission_encoder = LabelEncoder()

data["Fuel_Type"] = fuel_encoder.fit_transform(
    data["Fuel_Type"]
)

data["Seller_Type"] = seller_encoder.fit_transform(
    data["Seller_Type"]
)

data["Transmission"] = transmission_encoder.fit_transform(
    data["Transmission"]
)

print("\nEncoded Dataset:")
print(data.head())

# 12. Select Features

X = data[
    [
        "Year",
        "Present_Price",
        "Kms_Driven",
        "Fuel_Type",
        "Seller_Type",
        "Transmission",
        "Owner"
    ]
]

# Target variable
y = data["Selling_Price"]

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

# 13. Split Dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data:", X_train.shape)
print("Testing Data :", X_test.shape)

# 14. Create Random Forest Model

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# 15. Train Model

print("\nTraining Random Forest Model...")

model.fit(
    X_train,
    y_train
)

print("Training Completed!")

# ------------------------------------------------------------
# 16. Make Predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)

print("\nPredicted Prices:")
print(y_pred)

# ------------------------------------------------------------
# 17. Model Evaluation
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n")
print("=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print("Mean Absolute Error      :", round(mae, 3))
print("Mean Squared Error       :", round(mse, 3))
print("Root Mean Squared Error  :", round(rmse, 3))
print("R2 Score                 :", round(r2, 3))

# ------------------------------------------------------------
# 18. Actual vs Predicted Price Graph
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.8
)

# Perfect prediction line
minimum = min(
    y_test.min(),
    y_pred.min()
)

maximum = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Car Price")
plt.ylabel("Predicted Car Price")

plt.title(
    "Actual vs Predicted Car Prices"
)

plt.grid(True, alpha=0.3)

plt.show()

# ------------------------------------------------------------
# 19. Feature Importance
# ------------------------------------------------------------

importance = model.feature_importances_

feature_names = X.columns

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFeature Importance:")
print(feature_importance)

# ------------------------------------------------------------
# 20. Feature Importance Graph
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance in Car Price Prediction")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# 21. Actual vs Predicted Line Graph
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison = comparison.reset_index(drop=True)

plt.figure(figsize=(10, 6))

plt.plot(
    comparison["Actual"],
    marker="o",
    label="Actual Price"
)

plt.plot(
    comparison["Predicted"],
    marker="x",
    label="Predicted Price"
)

plt.xlabel("Test Car Number")
plt.ylabel("Selling Price (Lakhs)")

plt.title(
    "Actual Price vs Predicted Price"
)

plt.legend()

plt.grid(True, alpha=0.3)

plt.show()

# ------------------------------------------------------------
# 22. Predict Price of a New Car
# ------------------------------------------------------------

# Example new car:
# Year = 2018
# Present Price = 8.5 lakhs
# Kms Driven = 25000
# Fuel Type = Petrol
# Seller Type = Dealer
# Transmission = Manual
# Owner = 0

new_car = pd.DataFrame({
    "Year": [2018],
    "Present_Price": [8.5],
    "Kms_Driven": [25000],
    "Fuel_Type": [
        fuel_encoder.transform(["Petrol"])[0]
    ],
    "Seller_Type": [
        seller_encoder.transform(["Dealer"])[0]
    ],
    "Transmission": [
        transmission_encoder.transform(["Manual"])[0]
    ],
    "Owner": [0]
})

# Predict
new_prediction = model.predict(
    new_car
)

print("\n")
print("=" * 60)
print("NEW CAR PRICE PREDICTION")
print("=" * 60)

print(
    "Estimated Selling Price:",
    round(new_prediction[0], 2),
    "Lakhs"
)

# ------------------------------------------------------------
# 23. Final Result
# ------------------------------------------------------------

print("\n")
print("=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)

print("Algorithm Used : Random Forest Regression")
print("Number of Trees:", 100)
print("R2 Score       :", round(r2, 3))
print(
    "Predicted Price:",
    round(new_prediction[0], 2),
    "Lakhs"
)

# 1. Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import warnings
warnings.filterwarnings("ignore")

# 2. Load Dataset

data = pd.read_csv("car_price_dataset.csv")

print("=" * 60)
print("CAR PRICE PREDICTION")
print("=" * 60)

print("\nFirst 5 Records:")
print(data.head())

# 3. Display Dataset Information

print("\nDataset Shape:")
print(data.shape)

print("\nColumn Names:")
print(data.columns.tolist())

print("\nDataset Information:")
print(data.info())

print("\nStatistical Description:")
print(data.describe())

# 4. Check Missing Values

print("\nMissing Values:")
print(data.isnull().sum())

# 5. Check Duplicate Values

print("\nNumber of Duplicate Rows:")
print(data.duplicated().sum())

# Remove duplicate rows if present
data = data.drop_duplicates()

print("\nDataset Shape After Removing Duplicates:")
print(data.shape)

# 6. Exploratory Data Analysis

# Distribution of selling price

plt.figure(figsize=(9, 5))

plt.hist(
    data["Selling_Price"],
    bins=10,
    edgecolor="black"
)

plt.xlabel("Selling Price (Lakhs)")
plt.ylabel("Number of Cars")
plt.title("Distribution of Car Selling Prices")

plt.grid(axis="y", alpha=0.3)

plt.show()

# 7. Present Price vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Present_Price"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Present Price (Lakhs)")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Present Price vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 8. Year vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Year"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Manufacturing Year")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Year vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 9. Kms Driven vs Selling Price

plt.figure(figsize=(9, 5))

plt.scatter(
    data["Kms_Driven"],
    data["Selling_Price"],
    alpha=0.7
)

plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Kilometers Driven vs Selling Price")

plt.grid(True, alpha=0.3)

plt.show()

# 10. Fuel Type Analysis

plt.figure(figsize=(7, 5))

sns.boxplot(
    x="Fuel_Type",
    y="Selling_Price",
    data=data
)

plt.xlabel("Fuel Type")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Selling Price by Fuel Type")

plt.show()

# 11. Encode Categorical Variables

fuel_encoder = LabelEncoder()
seller_encoder = LabelEncoder()
transmission_encoder = LabelEncoder()

data["Fuel_Type"] = fuel_encoder.fit_transform(
    data["Fuel_Type"]
)

data["Seller_Type"] = seller_encoder.fit_transform(
    data["Seller_Type"]
)

data["Transmission"] = transmission_encoder.fit_transform(
    data["Transmission"]
)

print("\nEncoded Dataset:")
print(data.head())

# 12. Select Features

X = data[
    [
        "Year",
        "Present_Price",
        "Kms_Driven",
        "Fuel_Type",
        "Seller_Type",
        "Transmission",
        "Owner"
    ]
]

# Target variable
y = data["Selling_Price"]

print("\nInput Features:")
print(X.head())

print("\nTarget Variable:")
print(y.head())

# 13. Split Dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data:", X_train.shape)
print("Testing Data :", X_test.shape)

# 14. Create Random Forest Model

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

# 15. Train Model

print("\nTraining Random Forest Model...")

model.fit(
    X_train,
    y_train
)

print("Training Completed!")

# ------------------------------------------------------------
# 16. Make Predictions
# ------------------------------------------------------------

y_pred = model.predict(X_test)

print("\nPredicted Prices:")
print(y_pred)

# ------------------------------------------------------------
# 17. Model Evaluation
# ------------------------------------------------------------

mae = mean_absolute_error(
    y_test,
    y_pred
)

mse = mean_squared_error(
    y_test,
    y_pred
)

rmse = np.sqrt(mse)

r2 = r2_score(
    y_test,
    y_pred
)

print("\n")
print("=" * 60)
print("MODEL PERFORMANCE")
print("=" * 60)

print("Mean Absolute Error      :", round(mae, 3))
print("Mean Squared Error       :", round(mse, 3))
print("Root Mean Squared Error  :", round(rmse, 3))
print("R2 Score                 :", round(r2, 3))

# ------------------------------------------------------------
# 18. Actual vs Predicted Price Graph
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.8
)

# Perfect prediction line
minimum = min(
    y_test.min(),
    y_pred.min()
)

maximum = max(
    y_test.max(),
    y_pred.max()
)

plt.plot(
    [minimum, maximum],
    [minimum, maximum],
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Car Price")
plt.ylabel("Predicted Car Price")

plt.title(
    "Actual vs Predicted Car Prices"
)

plt.grid(True, alpha=0.3)

plt.show()

# ------------------------------------------------------------
# 19. Feature Importance
# ------------------------------------------------------------

importance = model.feature_importances_

feature_names = X.columns

feature_importance = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

print("\nFeature Importance:")
print(feature_importance)

# ------------------------------------------------------------
# 20. Feature Importance Graph
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    feature_importance["Feature"],
    feature_importance["Importance"]
)

plt.xlabel("Features")
plt.ylabel("Importance")
plt.title("Feature Importance in Car Price Prediction")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# 21. Actual vs Predicted Line Graph
# ------------------------------------------------------------

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison = comparison.reset_index(drop=True)

plt.figure(figsize=(10, 6))

plt.plot(
    comparison["Actual"],
    marker="o",
    label="Actual Price"
)

plt.plot(
    comparison["Predicted"],
    marker="x",
    label="Predicted Price"
)

plt.xlabel("Test Car Number")
plt.ylabel("Selling Price (Lakhs)")

plt.title(
    "Actual Price vs Predicted Price"
)

plt.legend()

plt.grid(True, alpha=0.3)

plt.show()

# ------------------------------------------------------------
# 22. Predict Price of a New Car
# ------------------------------------------------------------

# Example new car:
# Year = 2018
# Present Price = 8.5 lakhs
# Kms Driven = 25000
# Fuel Type = Petrol
# Seller Type = Dealer
# Transmission = Manual
# Owner = 0

new_car = pd.DataFrame({
    "Year": [2018],
    "Present_Price": [8.5],
    "Kms_Driven": [25000],
    "Fuel_Type": [
        fuel_encoder.transform(["Petrol"])[0]
    ],
    "Seller_Type": [
        seller_encoder.transform(["Dealer"])[0]
    ],
    "Transmission": [
        transmission_encoder.transform(["Manual"])[0]
    ],
    "Owner": [0]
})

# Predict
new_prediction = model.predict(
    new_car
)

print("\n")
print("=" * 60)
print("NEW CAR PRICE PREDICTION")
print("=" * 60)

print(
    "Estimated Selling Price:",
    round(new_prediction[0], 2),
    "Lakhs"
)

# ------------------------------------------------------------
# 23. Final Result
# ------------------------------------------------------------

print("\n")
print("=" * 60)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 60)

print("Algorithm Used : Random Forest Regression")
print("Number of Trees:", 100)
print("R2 Score       :", round(r2, 3))
print(
    "Predicted Price:",
    round(new_prediction[0], 2),
    "Lakhs"
)

```
